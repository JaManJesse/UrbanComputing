```python
import numpy as np
from scipy.stats import pearsonr  
```

```python
pred = np.load("../results/AGCRN/ILISTATES_pred.npy")
true = np.load("../results/AGCRN/ILISTATES_true.npy")
```

```python
# Pearson correlation per horizon
num_horizons = pred.shape[1]
per_horizon_stats = []
for h in range(num_horizons):
    pred_h = pred[:, h, ...].reshape(-1)
    true_h = true[:, h, ...].reshape(-1)
    mask_h = np.isfinite(pred_h) & np.isfinite(true_h)
    pred_h_clean = pred_h[mask_h]
    true_h_clean = true_h[mask_h]
    if pred_h_clean.size < 2:
        per_horizon_stats.append((np.nan, np.nan))
        continue
    r_h, p_h = pearsonr(pred_h_clean, true_h_clean)
    per_horizon_stats.append((r_h, p_h))

for idx, (r_h, p_h) in enumerate(per_horizon_stats, start=1):
    if np.isnan(r_h):
        print(f"Horizon {idx:02d}: insufficient data")
    else:
        print(f"Horizon {idx:02d}: r = {r_h:.4f}, p = {p_h:.3e}")
```
