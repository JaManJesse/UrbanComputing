```python
import pandas as pd
import numpy as np
```

```python
data = pd.read_csv("ILINet.csv", skiprows=1)
```

```python
data = data[["REGION", "YEAR", "WEEK", "ILITOTAL"]]
```

```python
data[:5]
```

```python
used_regions = (data[:100]["REGION"].unique()).tolist()
```

```python
print(len(used_regions))
```

```python
df = [data[data["REGION"] == region] for region in used_regions]
```

```python
for i in range(len(df)):
    region_df = df[i].copy()
    region_df["ILITOTAL"] = pd.to_numeric(region_df["ILITOTAL"], errors="coerce")
    region_df["YEAR"] = pd.to_numeric(region_df["YEAR"], errors="coerce")
    df[i] = region_df
```

```python
full_df = pd.concat(df, ignore_index=True)
full_df = full_df[full_df["YEAR"] < 2018]
```

```python
full_df.head()
```

```python
print(max(full_df["ILITOTAL"]))
```

```python
pivot_table = full_df.pivot_table(index=["YEAR", "WEEK"], columns="REGION", values="ILITOTAL")
```

```python
tensor = pivot_table.values[:,:,None]
np.save("us_infections", tensor)
```
