```python
import pandas as pd
```

```python
data = pd.read_csv("ILINet.csv", skiprows=1)
```

```python
data = data[["REGION", "YEAR", "WEEK", "ILITOTAL", "TOTAL PATIENTS"]]
```

```python
data[:5]
```

```python
used_regions = (data[:100]["REGION"].unique()).tolist()
```

```python
df = [data[data["REGION"] == region] for region in used_regions]
```

```python
for i in range(len(df)):
    region_df = df[i].copy()
    region_df["TOTAL PATIENTS"] = pd.to_numeric(region_df["TOTAL PATIENTS"], errors="coerce")
    region_df["NEW_PATIENTS"] = region_df["TOTAL PATIENTS"].diff()
    region_df["NEW_PATIENTS"] = region_df["NEW_PATIENTS"].fillna(0)
    df[i] = region_df
```

```python
full_df = pd.concat(df, ignore_index=True)
```
